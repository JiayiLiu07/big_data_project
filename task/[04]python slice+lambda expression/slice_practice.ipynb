### 2\. 练习题目

#### 题目 1：列表反转
给定一个列表，使用切片操作将其反转。
**示例输入**：`[1, 2, 3, 4, 5]`
**示例输出**：`[5, 4, 3, 2, 1]`

#### 题目 2：提取特定元素
给定一个列表，提取其中索引为奇数的元素。
**示例输入**：`[10, 20, 30, 40, 50, 60]`
**示例输出**：`[20, 40, 60]`

#### 题目 3：子列表提取
给定一个列表，提取从第 4 个元素开始到倒数第 3 个元素结束的子列表。
**示例输入**：`[1, 2, 3, 4, 5, 6, 7, 8, 9]`
**示例输出**：`[4, 5, 6]`

#### 题目 4：列表分隔
给定一个列表，每隔 2 个元素插入一个分隔符（例如 `None`）。
**示例输入**：`[1, 2, 3, 4, 5, 6]`
**示例输出**：`[1, 2, None, 3, 4, None, 5, 6]`

#### 题目 5：列表截取
给定一个列表，截取从第 3 个元素开始到第 7 个元素结束的子列表（包含第 7 个元素）。
**示例输入**：`[1, 2, 3, 4, 5, 6, 7,8, 9]`
**示例输出**：`[3, 4, 5, 6, 7]`

```python
#### 题目 1：列表反转
f=[1, 2, 3, 4, 5]
print(f[::-1])
#### 题目 2：提取特定元素
g=[10, 20, 30, 40, 50, 60]
print([g[i] for i in range(0,len(g)) if i%2!=0])
print(g[1::2])
#### 题目 3：子列表提取
h=[1, 2, 3, 4, 5, 6, 7, 8, 9]
print(h[3:-3:])

#### 题目 4：列表分隔
j=[1, 2, 3, 4, 5, 6]
print("None".join([j[i:i+3] for i in range(0,len(j),2)])) 

lst=[]
for x in range(0,len(j),2):
    lst.extend(j[x:x+2])
    if x+2<len(j):
        lst.append(None)
print(lst)

#### 题目 5：列表截取
k=[1, 2, 3, 4, 5, 6, 7,8, 9]
print(k[2:7])
```
